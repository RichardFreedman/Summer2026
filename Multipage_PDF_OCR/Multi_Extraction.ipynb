```python
from paddleocr import PaddleOCR

ocr = PaddleOCR(use_angle_cls=True, lang='en', enable_mkldnn=False)
```

```python
import threading
import pymupdf
import base64
import os

def extract_page(path,num):
    print("thread "+str(num)+" starting")
    result = ocr.predict(path)
    outp=[num,[]]
    for item in result:
        for text in item.get('rec_texts', []):
            outp[1].append(text)
    output.append(outp)
    print("thread "+str(num)+" ending")
    return
document=pymupdf.open('House of Leaves-5-10.pdf')
for i,page in enumerate(document):
    img=page.get_pixmap()
    img.save("/Users/holden/Downloads/Text Extractor/pic_dump/page"+str(i)+".png")
document.close()
output=[]
threads=[]
inpaths=[]
for dirpath, dirnames, filenames in os.walk("/Users/holden/Downloads/Text Extractor/pic_dump"):
    for file in filenames:
        if file[-4:]==".png":
            inpaths.append("/Users/holden/Downloads/Text Extractor/pic_dump/"+file)
print(inpaths)
for i,pth in enumerate(inpaths):
    t=threading.Thread(target=extract_page,args=(pth,i))
    threads.append(t)
for t in threads:
    t.start()
for t in threads:
    t.join()
print(output)
```
